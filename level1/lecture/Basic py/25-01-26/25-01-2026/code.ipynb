```python
def bill_split(no_per=1):
    total_amount = []

    while True:
        item_price = int(input("Enter price of item (0 to finish): "))

        if item_price == 0:
            break

        total_amount.append(item_price)

    total = sum(total_amount)
    tax_amount = total * 0.05
    tip_amount = total * 0.10
    net_amount = total + tax_amount + tip_amount
    each_person_cost = net_amount / no_per

    print(f"Total bill: {total}")
    print(f"Tax: {tax_amount}")
    print(f"Tip: {tip_amount}")
    print(f"Net amount: {net_amount}")
    print(f"Each person pays: {each_person_cost}")
```
