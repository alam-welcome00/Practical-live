1. The Office Door Badge You are programming a smart door for a tech office. The door should only open if the
user provides the correct "Access Code."

3. The Smart Home "Leaving for Work" Checklist When you leave your house, a smart assistant checks multiple
independent things. These are not mutually exclusive.
Input: Three separate inputs (Yes/No) for: "Is the AC on?", "Is the Coffee Maker on?", "Is the Front Door
unlocked?"
Scenario:
If AC is on Print "Turning off AC to save energy."
If Coffee Maker is on Print "Warning: Coffee Maker is still hot!"
If Door is unlocked Print "Locking the front door now." (Note: All three messages could potentially print.)

```python
ac_on = input("Is Ac on (Yes/No): ")
coffe_maker_on = input("Is Coffe maker working (Yes/No): ")
front_door_open = input("Is Fron door open (Yes/No)")


if ac_on == "Yes":
    print("Please switch of AC, it is still working")
else:
    print("Ac is off")
if coffe_maker_on == "Yes":
    print("Coffe maker is still on")
else:
    print("Coffe Maker is off")
if front_door_open == "Yes":
    print("Front door is open")
else:
    print("Front Door is closed")

if ac_on != "Yes" and coffe_maker_on != "Yes" and front_door_open!= "Yes":
    print("Every thing is fine now, you can leave")
else:
    print("Please of electronics and doors")
```

The Pizza Topping Calculator A customer is choosing toppings. Each topping adds a specific cost regardless of
the other toppings chosen.
Input: Three boolean (True/False) inputs for: extra_cheese , pepperoni , and olives .
Scenario: Initialize a total_bill = 10 .
If extra_cheese is True add 2 to bill.
If pepperoni is True add 3 to bill.
If olives is True add 1 to bill.
Output: Print the final total bill.


```python
total_bill = 10

pepperoni = input("Do you wants some pepperoni: ")
olives = input("Do you want olives: ")
extra_chess = input("Do you wants some extra chess: ")

if extra_chess == "Yes":
    total_bill+=2
if olives == "Yes":
    total_bill+=1
if pepperoni == "Yes":
    total_bill+=3

print(f"Your total bill is {total_bill}")
```

Movie Theater Ticket Pricing A theater charges based on the age of the customer.
Input: User enters their age.
Scenario:
Age < 5 "Ticket is Free."
Age 5 to 17 "Child Ticket: $10."
Age 18 to 60 "Adult Ticket: $18."
Age > 60 "Senior Ticket: $12."

```python
age = int(input("Enter Your Age: "))

if age<=5:
    print("Your Entry is free")
elif 5<age<=17:
    print("Child ticket: $10")
elif 18 <=age<=60:
    print("Adult Ticket: $18")
else:
    print("Senior Ticket: $12")
```

7. The 7-Day Steps Tracker A fitness app wants to show your activity for the week.
Data: A list of steps: [4000, 7500, 10000, 2000, 8000, 12000, 5000] .
Scenario: Loop through the list and for each day, print: "Day [X]: You walked [Steps] steps."

```python
seven_day_setps= [4000, 7500, 10000, 2000, 8000, 12000, 5000]
no_of_days = 1

for i in seven_day_setps:
    print(f"On day {no_of_days}, the total number of steps is {i}")
    no_of_days+=1
```

. The Email "Bcc" System A marketing manager has a list of 5 client names.
Input: A list containing 5 names.
Scenario: Use a loop to "send" an email to each person by printing: "Sending personalized promo code to
[Name]..."

```python
names = []

no_of_name = int(input("Enter Total number of names: "))

for i in range(no_of_name):
    name = input("Enter Name: ")
    names.append(name)

for i in names:
    print(f"Hello {i}, how are you, your promo code is {i.upper()}")
```
