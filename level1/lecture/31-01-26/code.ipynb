```python
import os
```

```python
def path_joiner(operating_system, *argus):
    final_path = ""
    for i in argus:
        if operating_system == "window":
            final_path = final_path + i+"\\"
        if operating_system == "macos":
            final_path = final_path + i+"\\"
    final_path = final_path[:-1]
    return final_path
```

```python
path_joiner("window", os.getcwd(),"temp","dummy.txt")
```
