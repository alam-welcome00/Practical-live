```python
res_bmi = df.groupby("Response")["BMI"].mean()
res_age = df.groupby("Response")["Ins_Age"].mean()
res_ht = df.groupby("Response")["Ht"].mean()
res_WT = df.groupby("Response")["Wt"].mean()
print(res_age)
print(res_bmi)
print(res_ht)
print(res_WT)
```

```python
from scipy.stats import chi2_contingency
```

```python
table = pd.crosstab(df_for_drop["Medical_History_24"], df_for_drop["Medical_History_32"])

chi2, p_value, dof, expected = chi2_contingency(table)

print(chi2,p_value,dof)
```

```python
from scipy.stats import chi2_contingency

for col in ['Medical_History_24', 'Medical_History_32']:
    is_missing = df_for_drop[col].isnull()
    table = pd.crosstab(is_missing, df_for_drop['Response'])
    chi2, p, dof, expected = chi2_contingency(table)
    print(col, '→ chi2:', chi2, 'p:', p, 'dof:', dof)
```
